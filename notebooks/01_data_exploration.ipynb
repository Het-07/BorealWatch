```python
categorical_columns = [
    "SIZE_CLASS",
    "FIRE_ORIGIN",
    "GENERAL_CAUSE",
    "TRUE_CAUSE",
    "FIRE_TYPE",
    "FIRE_POSITION_ON_SLOPE",
    "FUEL_TYPE"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False).head(15))
```

```python
date_columns = [
    "FIRE_START_DATE",
    "DISCOVERED_DATE",
    "REPORTED_DATE",
    "DISPATCH_DATE",
    "ASSESSMENT_DATETIME",
    "IA_ARRIVAL_AT_FIRE_DATE",
    "FIRE_FIGHTING_START_DATE",
    "FIRST_BUCKET_DROP_DATE",
    "FIRST_BH_DATE",
    "FIRST_UC_DATE",
    "FIRST_TO_DATE",
    "FIRST_EX_DATE"
]

for column in date_columns:
    df[column] = pd.to_datetime(df[column], errors="coerce")

df[date_columns].dtypes
```

```python
numeric_columns = [
    "CURRENT_SIZE",
    "LATITUDE",
    "LONGITUDE",
    "TEMPERATURE",
    "RELATIVE_HUMIDITY",
    "WIND_SPEED"
]

df[numeric_columns].describe().T
```

```python
invalid_coordinates = df[
    (df["LATITUDE"] < 48) |
    (df["LATITUDE"] > 60) |
    (df["LONGITUDE"] > -110) |
    (df["LONGITUDE"] < -121)
]

print("Potentially invalid coordinates:", len(invalid_coordinates))
```
